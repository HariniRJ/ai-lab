```python
# 8-Puzzle Problem using Iterative Deepening Search (IDS)
# Student Details:
print(" HARINI R J")
print("1BM24CS364")

# Define the goal state (0 represents the empty space)
GOAL_STATE = (1, 2, 3, 4, 5, 6, 7, 8, 0)

# Move definitions with their 1D tuple offsets
MOVES = {
    'UP': -3,
    'DOWN': 3,
    'LEFT': -1,
    'RIGHT': 1
}

def get_successors(state):
    """Finds all valid next states by shifting the empty space (0)."""
    blank_idx = state.index(0)
    row = blank_idx // 3
    col = blank_idx % 3
    successors = []

    for move, offset in MOVES.items():
        # Prevent the blank space from wrapping around edges
        if move == 'UP' and row == 0: continue
        if move == 'DOWN' and row == 2: continue
        if move == 'LEFT' and col == 0: continue
        if move == 'RIGHT' and col == 2: continue

        # Generate the next state by swapping elements
        new_idx = blank_idx + offset
        new_state = list(state)
        new_state[blank_idx], new_state[new_idx] = new_state[new_idx], new_state[blank_idx]
        successors.append((tuple(new_state), move))

    return successors

def depth_limited_search(state, depth, path, visited):
    """Performs standard Depth-Limited Search (DLS) up to the limit."""
    if state == GOAL_STATE:
        return path

    if depth <= 0:
        return None

    # Track states visited along the current recursive tree path to avoid loops
    visited.add(state)

    for successor, move in get_successors(state):
        if successor not in visited:
            result = depth_limited_search(successor, depth - 1, path + [move], visited)
            if result is not None:
                return result

    # Backtrack: Allow other branches to traverse this state if needed
    visited.remove(state)
    return None

def iterative_deepening_search(start_state):
    """Loops and gradually expands the maximum search depth until a path is found."""
    depth = 0
    while True:
        visited = set()
        result = depth_limited_search(start_state, depth, [], visited)
        if result is not None:
            return result, depth
        depth += 1

def print_board(state):
    """Helper formatting function to output the 3x3 layout."""
    for i in range(0, 9, 3):
        print(f"[ {state[i]} {state[i+1]} {state[i+2]} ]")
    print()

# --- Execution ---
if __name__ == "__main__":


    # We use a reachable initial state located a few steps away.
    initial_state = (1, 2, 3, 0, 4, 6, 7, 5, 8)

    print("Initial Board State:")
    print_board(initial_state)

    print("Executing Iterative Deepening Search...")
    solution_path, max_depth = iterative_deepening_search(initial_state)

    if solution_path is not None:
        print(f"🎉 Goal reached at depth: {max_depth}")
        print(f"Moves taken ({len(solution_path)} steps): {', '.join(solution_path)}")
    else:
        print("No solution path could be generated.")
```
