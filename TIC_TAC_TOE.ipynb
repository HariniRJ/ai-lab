```python
import random

print ('harini','1BM24CS364')

board = [
    [' ', ' ', ' '],
    [' ', ' ', ' '],
    [' ', ' ', ' ']
]

winning = [
    [(0,0), (0,1), (0,2)],
    [(1,0), (1,1), (1,2)],
    [(2,0), (2,1), (2,2)],
    [(0,0), (1,0), (2,0)],
    [(0,1), (1,1), (2,1)],
    [(0,2), (1,2), (2,2)],
    [(0,0), (1,1), (2,2)],
    [(0,2), (1,1), (2,0)]
]


def check_winner(board):
    for line in winning:
        a, b, c = line

        if board[a[0]][a[1]] != ' ' and \
           board[a[0]][a[1]] == board[b[0]][b[1]] == board[c[0]][c[1]]:
            return board[a[0]][a[1]]

    return None


while True:

    # Human turn
    print(board)

    row = int(input("Enter row (0-2): "))
    col = int(input("Enter column (0-2): "))

    while board[row][col] != ' ':
        print("Position already occupied. Try again.")
        row = int(input("Enter row (0-2): "))
        col = int(input("Enter column (0-2): "))

    board[row][col] = 'O'

    winner = check_winner(board)

    if winner:
        print(board)
        print("Winner:", winner)
        break


    # AI turn
    empty_cells = []
    for r in range(3):
        for c in range(3):
            if board[r][c] == ' ':
                empty_cells.append((r, c))

    if empty_cells:
        ai_row, ai_col = random.choice(empty_cells)
        board[ai_row][ai_col] = 'X'

    winner = check_winner(board)

    if winner:
        print(board)
        print("Winner:", winner)
        break


    # Check draw
    full = True

    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                full = False
                break
        if not full:
            break

    if full:
        print(board)
        print("Game Draw!")
        break
```
