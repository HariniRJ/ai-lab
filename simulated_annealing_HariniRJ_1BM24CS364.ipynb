```python
import random
import math

def calculate_cost(state):
    """
    Calculates the number of attacking pairs of queens (cost).
    We want to minimize this value to 0.
    """
    attacks = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Same row attack
            if state[i] == state[j]:
                attacks += 1
            # Diagonal attack
            elif abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_random_neighbor(state):
    """
    Generates a random neighbor by moving one random queen to a different random row.
    """
    neighbor = list(state)
    col = random.randint(0, 7)
    row = random.randint(0, 7)

    # Ensure we actually move the queen to a new row
    while row == state[col]:
        row = random.randint(0, 7)

    neighbor[col] = row
    return neighbor

def simulated_annealing():
    # 1: current <- initial state (random row for each of the 8 columns)
    current = [random.randint(0, 7) for _ in range(8)]
    current_cost = calculate_cost(current)

    # 2: T <- a large positive value
    T = 1000.0
    cooling_rate = 0.995  # Used to gradually decrease T

    # 3: while T > 0 do (using a small threshold for floating-point safety)
    while T > 0.0001:
        if current_cost == 0:
            return current  # Found a perfect solution early!

        # 4: next <- a random neighbour of current
        neighbor = get_random_neighbor(current)
        neighbor_cost = calculate_cost(neighbor)

        # 5: ΔE <- current.cost - next.cost
        delta_E = current_cost - neighbor_cost

        # 6: if ΔE > 0 then
        if delta_E > 0:
            # 7: current <- next
            current = neighbor
            current_cost = neighbor_cost
        # 8: else
        else:
            # 9: current <- next with probability p = e^(ΔE / T)
            probability = math.exp(delta_E / T)
            if random.random() < probability:
                current = neighbor
                current_cost = neighbor_cost
        # 10: end if

        # 11: decrease T
        T *= cooling_rate

    # 12: end while
    # 13: return current
    return current

# --- Execution ---
if __name__ == "__main__":
    solution = simulated_annealing()
    final_cost = calculate_cost(solution)

    print(f"Final State (Rows per Column): {solution}")
    print(f"Remaining Attacking Pairs (Cost): {final_cost}")

    if final_cost == 0:
        print("\nSuccess! Found a valid 8-Queens solution:")
        # Visualizing the board
        for row in range(8):
            line = ""
            for col in range(8):
                if solution[col] == row:
                    line += " Q "
                else:
                    line += " . "
            print(line)
    else:
        print("\nStopped at a local optimum. Try running the script again!")
```
