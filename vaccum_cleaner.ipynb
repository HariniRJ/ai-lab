```python
room = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"

while room["A"] == "Dirty" or room["B"] == "Dirty":

    status = room[location]

    if status == "Dirty":
        print("Pick Dust")
        room[location] = "Clean"

    elif location == "A":
        print("Move Right")
        location = "B"

    elif location == "B":
        print("Move Left")
        location = "A"

print("Both rooms are clean!")
```
