```python
def tt_entails(kb, alpha, symbols):
    """
    Determines if KB entails alpha and prints the truth table.
    """
    symbol_list = sorted(list(symbols))

    # Define headers matching the example layout
    headers = symbol_list + ["A V C", "B V ¬C", "KB", "α", "KB ⊨ α?"]
    print(" | ".join(f"{h:^10}" for h in headers))
    print("-" * (13 * len(headers)))

    return tt_check_all(kb, alpha, symbol_list, {}, symbol_list)


def tt_check_all(kb, alpha, symbols, model, all_symbols):
    """
    Recursively evaluates all truth table rows and prints the values.
    """
    if not symbols:
        # Evaluate intermediate components and main sentences
        avc = model['A'] or model['C']
        bvnotc = model['B'] or (not model['C'])
        kb_val = pl_true(kb, model)
        alpha_val = pl_true(alpha, model)

        # Determine row validity: if KB is True, alpha must be True
        valid = alpha_val if kb_val else True

        # Prepare row data for printing
        row = [str(model[s]).lower() for s in all_symbols]
        row += [str(avc).lower(), str(bvnotc).lower(), str(kb_val).lower(), str(alpha_val).lower(), "ok" if valid else "FAILED"]
        print(" | ".join(f"{val:^10}" for val in row))

        return valid

    else:
        p = symbols[0]
        rest = symbols[1:]

        # Recurse: Assign False first, then True (to match standard truth table ordering)
        model_false = model.copy()
        model_false[p] = False
        res_false = tt_check_all(kb, alpha, rest, model_false, all_symbols)

        model_true = model.copy()
        model_true[p] = True
        res_true = tt_check_all(kb, alpha, rest, model_true, all_symbols)

        return res_false and res_true


def pl_true(sentence, model):
    return sentence(model)


# ==========================================
# TEST WITH THE IMAGE EXAMPLE
# ==========================================
if __name__ == "__main__":
    # α = A V B
    # KB = (A V C) ∧ (B V ¬C)
    alpha_query = lambda m: m['A'] or m['B']
    kb_sentence = lambda m: (m['A'] or m['C']) and (m['B'] or not m['C'])

    proposition_symbols = {'A', 'B', 'C'}

    print("\n--- ENUMERATION TRUTH TABLE ---")
    entails = tt_entails(kb_sentence, alpha_query, proposition_symbols)
    print("--------------------------------")
    print(f"\nFinal Result: Does KB ⊨ α? {entails}\n")
```
